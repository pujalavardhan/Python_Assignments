#### Assignment 3: Calculator Using Functions

```python
def add(num1, num2):
    return num1 + num2

def subtract(num1, num2):
    return num1 - num2

def multiply(num1, num2):
    return num1 * num2

def divide(num1, num2):
    if num2 == 0:
        return "Cannot divide by zero"
    return num1 / num2


num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

operation = input("Enter operation (add/subtract/multiply/divide): ").lower()

if operation == "add":
    result = add(num1, num2)

elif operation == "subtract":
    result = subtract(num1, num2)

elif operation == "multiply":
    result = multiply(num1, num2)

elif operation == "divide":
    result = divide(num1, num2)

else:
    result = "Invalid Operation"

print("Result:", result)
```

#### Assignment 19: Find Maximum in a List

```python
from functools import reduce
def find_list_maximum(lst):
    if not lst:
        return "List is empty"

    max_number = reduce(lambda x, y: x if x > y else y, lst)
    return max_number


numbers = list(map(int, input("Enter the numbers: ").split()))

result = find_list_maximum(numbers)

print("The max number is", result)


```

#### Assignment 20: Count Positive, Negative and Zero Values

```python
def count_number_types(numbers):
    positive = 0
    negative = 0
    zero = 0

    for num in numbers:
        if num > 0:
            positive += 1
        elif num < 0:
            negative += 1
        else:
            zero += 1

    return positive, negative, zero


numbers = list(map(int, input("Enter numbers: ").split()))

positive_count, negative_count, zero_count = count_number_types(numbers)

print("Positive numbers:", positive_count)
print("Negative numbers:", negative_count)
print("Zeros:", zero_count)
```

 #### Assignment 28: Positional-Only Parameters

```python
def divide_numbers(a, b, /):
    # a and b are positional-only because of the '/' marker
    if b == 0:
        return "Error: Cannot divide by zero"
    return a / b


num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Correct call - using positional arguments
result = divide_numbers(num1, num2)
print("Result:", result)

# Incorrect call using keyword arguments (commented out on purpose)
# divide_numbers(a=num1, b=num2)
# The line above raises:
# TypeError: divide_numbers() got some positional-only arguments passed as keyword arguments: 'a, b'
# This happens because a and b are declared before the '/' in the function
# definition, which forces the caller to pass them only by position, not by name.
```

#### Assignment 30: Mixed Parameter Types

```python
def generate_report(report_id, /, title, status="Draft", *tags, author, **details):
    # report_id            -> positional-only parameter (before '/')
    # title                -> normal parameter (positional or keyword)
    # status               -> default parameter
    # *tags                -> variable positional arguments
    # author               -> keyword-only parameter (after *tags)
    # **details            -> variable keyword arguments
    print("Report Generator")
    print("----------------")
    print("Report ID:", report_id)
    print("Title    :", title)
    print("Status   :", status)
    print("Tags     :", tags)
    print("Author   :", author)
    print("Extra Details:")
    for key, value in details.items():
        print(f"  {key}: {value}")


# Demonstrating all parameter types in a single call
generate_report(
    101,                     # report_id (positional-only)
    "Sales Summary",         # title (normal)
    "Final",                 # status (overrides default)
    "monthly", "sales",      # *tags
    author="Naga",           # keyword-only
    department="Finance",    # extra keyword argument -> goes into **details
    pages=5                  # extra keyword argument -> goes into **details
)
```

#### Assignment 31: Return Multiple Values

#### Assignment 32: Function Returning a Dictionary

```python
def analyze_text(text):
    vowels = "aeiouAEIOU"

    total_characters = len(text)
    total_words = len(text.split())
    total_vowels = 0
    total_digits = 0
    total_spaces = 0

    for ch in text:
        if ch in vowels:
            total_vowels += 1
        elif ch.isdigit():
            total_digits += 1
        elif ch == " ":
            total_spaces += 1

    return {
        "Total Characters": total_characters,
        "Total Words": total_words,
        "Total Vowels": total_vowels,
        "Total Digits": total_digits,
        "Total Spaces": total_spaces
    }


sentence = input("Enter a sentence: ")
report = analyze_text(sentence)

print("\nText Analysis")
print("-------------")
for key, value in report.items():
    print(f"{key}: {value}")
```

#### Assignment 33: Function Returning a List

```python
def get_even_numbers(start, end):
    if start >= end:
        return "Error: Starting number must be smaller than ending number"

    even_numbers = []
    for num in range(start, end + 1):
        if num % 2 == 0:
            even_numbers.append(num)

    return even_numbers


start_num = int(input("Enter starting number: "))
end_num = int(input("Enter ending number: "))

result = get_even_numbers(start_num, end_num)
print("Even numbers:", result)
```

#### Assignment 34: Prime Number Function

```python
def is_prime(number):
    # A prime number has exactly two factors: 1 and itself
    if number < 2:
        return False
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            return False
    return True


print("Prime numbers from 1 to 100:")
primes = []
for n in range(1, 101):
    if is_prime(n):
        primes.append(n)

print(primes)
```

#### Assignment 35: Fibonacci Function

```python
def generate_fibonacci(terms):
    if terms <= 0:
        return []
    if terms == 1:
        return [0]

    series = [0, 1]
    while len(series) < terms:
        next_term = series[-1] + series[-2]
        series.append(next_term)

    return series


num_terms = int(input("Enter number of terms: "))
result = generate_fibonacci(num_terms)
print(result)
```

#### Assignment 56: Mask Email Addresses

```python
def mask_email(email):
    username, domain = email.split("@")
    if len(username) <= 1:
        masked_username = username
    else:
        masked_username = username[0] + "*" * (len(username) - 1)
    return masked_username + "@" + domain


emails = ["bharath@gmail.com", "naga.kumar@yahoo.com", "a@test.com"]

masked_emails = list(map(mask_email, emails))

for original, masked in zip(emails, masked_emails):
    print(f"{original}  ->  {masked}")
```

#### Assignment 92: Closure-Based Bank Balance

```python
def create_bank_account(initial_balance=0):
    balance = initial_balance

    def deposit(amount):
        nonlocal balance
        balance += amount
        return balance

    def withdraw(amount):
        nonlocal balance
        if amount > balance:
            return "Error: Insufficient balance"
        balance -= amount
        return balance

    def check_balance():
        return balance

    return deposit, withdraw, check_balance


deposit, withdraw, check_balance = create_bank_account(1000)

print("Initial balance:", check_balance())
print("After deposit of 500:", deposit(500))
print("After withdrawal of 300:", withdraw(300))
print("After withdrawal of 5000:", withdraw(5000))
print("Final balance:", check_balance())
```

#### Assignment 100: Complete Functional Programming Project

```python
from functools import reduce

# ---------------------------------------------------------
# Sample Employee Data
# ---------------------------------------------------------
employees = [
    {"id": 101, "name": "Ravi", "department": "IT", "salary": 60000, "performance_score": 88, "experience": 5},
    {"id": 102, "name": "Anitha", "department": "HR", "salary": 45000, "performance_score": 72, "experience": 4},
    {"id": 103, "name": "Kiran", "department": "IT", "salary": 75000, "performance_score": 94, "experience": 7},
    {"id": 104, "name": "Meena", "department": "Finance", "salary": 55000, "performance_score": 81, "experience": 6},
    {"id": 105, "name": "Arjun", "department": "IT", "salary": 40000, "performance_score": 65, "experience": 2},
]


# ---------------------------------------------------------
# Data validation (salary and performance score must be sensible)
# ---------------------------------------------------------
def validate_employees(emp_list):
    valid_employees = []
    for emp in emp_list:
        if emp["salary"] < 0:
            print(f"Skipping {emp['name']}: invalid (negative) salary")
            continue
        if not (0 <= emp["performance_score"] <= 100):
            print(f"Skipping {emp['name']}: invalid performance score")
            continue
        valid_employees.append(emp)
    return valid_employees


# ---------------------------------------------------------
# Part 1: Display Employees
# ---------------------------------------------------------
def display_employees(emp_list):
    if not emp_list:
        print("No employees to display.")
        return
    for emp in emp_list:
        print(f"ID: {emp['id']} | Name: {emp['name']} | Dept: {emp['department']} | "
              f"Salary: {emp['salary']} | Performance: {emp['performance_score']} | "
              f"Experience: {emp['experience']}")


# ---------------------------------------------------------
# Part 2: Extract Employee Names (map)
# ---------------------------------------------------------
def get_employee_names(emp_list):
    return list(map(lambda emp: emp["name"], emp_list))


# ---------------------------------------------------------
# Part 3: Filter High Performers (filter)
# ---------------------------------------------------------
def get_high_performers(emp_list):
    return list(filter(lambda emp: emp["performance_score"] >= 80, emp_list))


# ---------------------------------------------------------
# Part 4: Filter by Department (filter)
# ---------------------------------------------------------
def filter_by_department(emp_list, department):
    return list(filter(lambda emp: emp["department"].lower() == department.lower(), emp_list))


# ---------------------------------------------------------
# Part 5: Salary Increase (map) - does not modify original dictionaries
# ---------------------------------------------------------
def get_increase_percent(score):
    if score >= 90:
        return 15
    elif score >= 80:
        return 10
    elif score >= 70:
        return 5
    else:
        return 0


def calculate_revised_salaries(emp_list):
    def apply_increase(emp):
        increase_percent = get_increase_percent(emp["performance_score"])
        revised_salary = round(emp["salary"] + (emp["salary"] * increase_percent / 100), 2)
        # {**emp, ...} creates a brand-new dictionary instead of editing 'emp' directly
        return {**emp, "revised_salary": revised_salary}

    return list(map(apply_increase, emp_list))


# ---------------------------------------------------------
# Part 6: Total Salary Expense (reduce)
# ---------------------------------------------------------
def calculate_salary_expense(emp_list, revised_list):
    current_total = reduce(lambda acc, emp: acc + emp["salary"], emp_list, 0)
    revised_total = reduce(lambda acc, emp: acc + emp["revised_salary"], revised_list, 0)
    additional_expense = round(revised_total - current_total, 2)
    return current_total, revised_total, additional_expense


# ---------------------------------------------------------
# Part 7: Highest-Paid Employee (reduce, no max())
# ---------------------------------------------------------
def get_highest_paid(emp_list):
    return reduce(lambda a, b: a if a["salary"] > b["salary"] else b, emp_list)


# ---------------------------------------------------------
# Part 8: Sort Employees (sorted + lambda)
# ---------------------------------------------------------
def sort_employees(emp_list, sort_by="name"):
    sort_keys = {
        "name": lambda emp: emp["name"],
        "salary": lambda emp: emp["salary"],
        "performance": lambda emp: emp["performance_score"],
        "experience": lambda emp: emp["experience"],
    }
    key_function = sort_keys.get(sort_by, sort_keys["name"])
    return sorted(emp_list, key=key_function)


# ---------------------------------------------------------
# Part 9: Closure-Based Bonus Calculator
# ---------------------------------------------------------
def create_bonus_calculator(bonus_percent):
    def calculate_bonus(salary):
        return round(salary * bonus_percent / 100, 2)

    return calculate_bonus


bonus_5_percent = create_bonus_calculator(5)
bonus_10_percent = create_bonus_calculator(10)
bonus_15_percent = create_bonus_calculator(15)


# ---------------------------------------------------------
# Part 10 & 11: Nested Report Generator + LEGB Demonstration
# ---------------------------------------------------------
company_name = "Edukron"  # Global scope variable


def generate_department_report(department, emp_list):
    # 'department' (parameter) and 'report_title' (below) become ENCLOSING scope
    # variables for the nested functions defined inside this function.
    report_title = f"{department} Department Report"
    dept_employees = list(filter(lambda emp: emp["department"].lower() == department.lower(), emp_list))

    def count_employees():
        # 'dept_employees' is resolved from the Enclosing scope (E in LEGB)
        return len(dept_employees)

    def total_salary():
        return reduce(lambda acc, emp: acc + emp["salary"], dept_employees, 0)

    def average_salary():
        count = count_employees()  # 'count' here is LOCAL (L in LEGB) to this function
        return round(total_salary() / count, 2) if count > 0 else 0

    def average_performance():
        count = count_employees()
        if count == 0:
            return 0
        total_score = reduce(lambda acc, emp: acc + emp["performance_score"], dept_employees, 0)
        return round(total_score / count, 2)

    # LEGB resolution used while building the report below:
    #   company_name  -> GLOBAL scope     (defined at module level)
    #   department    -> ENCLOSING scope  (parameter of generate_department_report)
    #   report_title  -> ENCLOSING scope  (local variable of generate_department_report)
    #   len, round    -> BUILT-IN scope   (Python's built-in functions)
    report = {
        "company": company_name,
        "title": report_title,
        "department": department,
        "employee_count": count_employees(),
        "total_salary": total_salary(),
        "average_salary": average_salary(),
        "average_performance_score": average_performance(),
    }
    return report


# ---------------------------------------------------------
# Part 12: Menu-Driven Program
# ---------------------------------------------------------
def display_menu():
    print("\n===== Employee Performance Analysis System =====")
    print("1. Display all employees")
    print("2. Display employee names")
    print("3. Display high performers")
    print("4. Search by department")
    print("5. Calculate revised salaries")
    print("6. Calculate total salary expense")
    print("7. Find highest-paid employee")
    print("8. Sort employees")
    print("9. Generate department report")
    print("10. Exit")


def run_menu(emp_list):
    if not emp_list:
        print("The employee list is empty. Nothing to process.")
        return

    while True:
        display_menu()
        choice = input("Enter your choice (1-10): ").strip()

        if choice == "1":
            display_employees(emp_list)

        elif choice == "2":
            print("Employee Names:", get_employee_names(emp_list))

        elif choice == "3":
            print("High Performers (score >= 80):")
            display_employees(get_high_performers(emp_list))

        elif choice == "4":
            department = input("Enter department name: ").strip()
            matched = filter_by_department(emp_list, department)
            if not matched:
                print(f"No employees found in the '{department}' department.")
            else:
                display_employees(matched)

        elif choice == "5":
            revised = calculate_revised_salaries(emp_list)
            for emp in revised:
                print(f"{emp['name']}: {emp['salary']} -> {emp['revised_salary']}")

        elif choice == "6":
            revised = calculate_revised_salaries(emp_list)
            current_total, revised_total, additional = calculate_salary_expense(emp_list, revised)
            print("Current total salary :", current_total)
            print("Revised total salary :", revised_total)
            print("Additional expense    :", additional)

        elif choice == "7":
            top_employee = get_highest_paid(emp_list)
            print("Highest-paid employee:", top_employee["name"], "-", top_employee["salary"])

        elif choice == "8":
            sort_field = input("Sort by (name/salary/performance/experience): ").strip().lower()
            if sort_field not in ("name", "salary", "performance", "experience"):
                print("Invalid field entered, defaulting to 'name'.")
                sort_field = "name"
            display_employees(sort_employees(emp_list, sort_field))

        elif choice == "9":
            department = input("Enter department for the report: ").strip()
            report = generate_department_report(department, emp_list)
            if report["employee_count"] == 0:
                print(f"No employees found in the '{department}' department.")
            else:
                for key, value in report.items():
                    print(f"{key}: {value}")

        elif choice == "10":
            print("Exiting the Employee Performance Analysis System. Goodbye!")
            break

        else:
            print("Invalid choice. Please enter a number between 1 and 10.")


# ---------------------------------------------------------
# Program entry point
# ---------------------------------------------------------
employees = validate_employees(employees)

print("Closure-based bonus calculators (on a salary of 60000):")
print("5% bonus :", bonus_5_percent(60000))
print("10% bonus:", bonus_10_percent(60000))
print("15% bonus:", bonus_15_percent(60000))

run_menu(employees)

# ---------------------------------------------------------
# Concept Summary
# ---------------------------------------------------------
# Functions        : Reusable, named blocks of code that take inputs (parameters)
#                     and can send back a result (return value).
# Lambda expressions: Small, unnamed, single-expression functions - handy for
#                     short operations passed directly into map(), filter(), sorted(), etc.
# map()            : Applies a function to every item of an iterable and returns
#                     the transformed items (e.g. revised salaries above).
# filter()         : Applies a boolean-returning function to every item and keeps
#                     only the items for which it returns True (e.g. high performers).
# reduce()         : Repeatedly combines all items of an iterable into a single
#                     cumulative value (e.g. total salary expense).
# Nested functions : Functions defined inside another function; useful for
#                     organizing logic that only makes sense within that context
#                     (e.g. the helper functions inside generate_department_report()).
# Closures         : A nested function that "remembers" variables from its
#                     enclosing function even after that function has finished
#                     running (e.g. the bonus calculators and bank account example).
# Local scope      : Names created inside a function; only visible inside it.
# Enclosing scope  : Names that belong to an outer function, visible to the
#                     nested function(s) defined within it.
# Global scope     : Names defined at the top level of the module; visible
#                     everywhere in the file unless shadowed by a local name.
# Built-in scope    : Names provided by Python itself (len, print, round, etc.),
#                     checked last in the LEGB lookup order.
# Normal function vs lambda: A normal function (def) can have a name, multiple
#                     statements, and a docstring, which makes it suited to
#                     complex, reusable logic. A lambda is restricted to a
#                     single expression and is usually anonymous, so it works
#                     best for small, throwaway operations passed as arguments.
```
