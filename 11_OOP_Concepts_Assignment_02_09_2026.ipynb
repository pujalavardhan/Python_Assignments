### Q8. Protect a amount with a property for banking case 1; use sample reference `ACC-1008` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
class Transfer:
    def __init__(self, amount):
        self._amount = None
        self.amount = amount   # goes through the setter below

    @property
    def amount(self):
        return self._amount

    @amount.setter
    def amount(self, value):
        if value <= 0:
            raise ValueError("Transfer amount must be positive")
        self._amount = value

t = Transfer(2500)
print(f"Transfer amount set to {t.amount}")

try:
    t.amount = -100
except ValueError as e:
    print(f"Could not update amount: {e}")

# Explanation: a property lets us keep the validation logic (amount > 0)
# while still using the simple "t.amount" syntax instead of a get/set method.
```

### Q9. Create portfolio and add_item for banking case 1; use sample reference `ACC-1009` and explain the result.

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
class Portfolio:
    def __init__(self, owner):
        self.owner = owner
        self.items = []

    def add_item(self, item_name, value):
        self.items.append((item_name, value))

    def total_value(self):
        return sum(value for _, value in self.items)

portfolio = Portfolio("Ananya Rao")
portfolio.add_item("Fixed Deposit", 50000)
portfolio.add_item("Mutual Fund - Growth", 15000)

print(f"Portfolio for {portfolio.owner}: {portfolio.items}")
print(f"Total portfolio value: {portfolio.total_value()}")

# Explanation: Portfolio "has-a" list of items (composition); add_item keeps
# the list-management logic inside the class instead of scattered outside it.
```

### Q12. Use a class method as an alternate constructor for banking case 1; use sample reference `ACC-1012` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
class Account:
    def __init__(self, account_ref, holder_name, balance):
        self.account_ref = account_ref
        self.holder_name = holder_name
        self.balance = balance

    @classmethod
    def from_new_customer(cls, account_ref, holder_name):
        # a brand new customer always starts with zero balance
        return cls(account_ref, holder_name, balance=0)

acc = Account.from_new_customer("ACC-1012", "Ananya Rao")
print(f"New account created via classmethod: {acc.account_ref}, balance = {acc.balance}")

# Explanation: a classmethod is an alternate way to build an object (here,
# always starting new customers at zero balance) without repeating logic.
```

### Q14. Model a loyaltyaccount for banking case 1; use sample reference `ACC-1014` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
class LoyaltyAccount:
    def __init__(self, account_ref, points=0):
        self.account_ref = account_ref
        self.points = points

    def add_points(self, points_earned):
        self.points += points_earned

    def redeem_points(self, points_needed):
        if points_needed > self.points:
            raise ValueError("Not enough loyalty points to redeem")
        self.points -= points_needed

loyalty = LoyaltyAccount("ACC-1014", points=1200)
loyalty.add_points(300)
print(f"Points after earning: {loyalty.points}")

try:
    loyalty.redeem_points(999999)
except ValueError as e:
    print(f"Redeem failed: {e}")

# Explanation: LoyaltyAccount keeps the points balance and the rules for
# changing it (add/redeem) together in one place.
```

```python
class PricingRule:
    def apply(self, amount):
        raise NotImplementedError

class FlatFeeRule(PricingRule):
    def __init__(self, fee):
        self.fee = fee

    def apply(self, amount):
        return amount + self.fee

class PercentageFeeRule(PricingRule):
    def __init__(self, percent):
        self.percent = percent

    def apply(self, amount):
        return amount + (amount * self.percent / 100)

class PricingEngine:
    def __init__(self, rules):
        self.rules = rules   # list of PricingRule objects, easy to extend

    def calculate_total(self, amount):
        total = amount
        for rule in self.rules:
            total = rule.apply(total)
        return total

engine = PricingEngine([FlatFeeRule(10), PercentageFeeRule(2)])
final_amount = engine.calculate_total(2500)
print(f"Final chargeable amount for 2500: {final_amount}")

# Explanation: new pricing rules can be added just by writing a new class
# that follows the same apply(amount) interface, without changing PricingEngine.
```

### Q95. Implement a unit-of-work transaction for banking case 3; use sample reference `ACC-1095` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
class UnitOfWork:
    def __init__(self, accounts):
        self.accounts = accounts
        self.changes = {}   # pending changes, not applied yet

    def __enter__(self):
        self.changes = {}
        return self

    def update_balance(self, account_ref, new_balance):
        self.changes[account_ref] = new_balance

    def __exit__(self, exc_type, exc_value, traceback):
        if exc_type is None:
            # no error happened, so commit all pending changes at once
            self.accounts.update(self.changes)
            print(f"Committed changes: {self.changes}")
        else:
            print(f"Rolled back, no changes applied. Reason: {exc_value}")
        return True

accounts = {"ACC-1095": 22000}

with UnitOfWork(accounts) as uow:
    uow.update_balance("ACC-1095", 22000 - 4300)
    uow.update_balance("ACC-9999", 5000)

print(f"Accounts after successful unit of work: {accounts}")

with UnitOfWork(accounts) as uow:
    uow.update_balance("ACC-1095", 0)
    raise ValueError("Simulated failure mid-transaction")

print(f"Accounts after failed unit of work (unchanged): {accounts}")

# Explanation: all changes are collected first and only written to the real
# accounts dictionary if nothing goes wrong (all-or-nothing, like a database transaction).
```

### Q96. Separate domain and infrastructure layers for banking case 3; use sample reference `ACC-1096` and explain the result.

```python
# Domain layer: pure business logic, does not know about files/databases
class Account:
    def __init__(self, account_ref, balance):
        self.account_ref = account_ref
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("Insufficient balance")
        self.balance -= amount

# Infrastructure layer: handles storage details (here, just a dictionary
# standing in for a real database)
class InMemoryAccountStore:
    def __init__(self):
        self._data = {}

    def save(self, account):
        self._data[account.account_ref] = account.balance

    def load_balance(self, account_ref):
        return self._data.get(account_ref)

account = Account("ACC-1096", 22000)
account.withdraw(4300)

store = InMemoryAccountStore()
store.save(account)

print(f"Domain object balance: {account.balance}")
print(f"Infrastructure stored balance: {store.load_balance('ACC-1096')}")

# Explanation: Account (domain) only knows banking rules, while
# InMemoryAccountStore (infrastructure) only knows how to save/load data;
# neither class depends on the other's internal details.
```

```python
class PaymentPlugin:
    def process(self, amount):
        raise NotImplementedError

class UPIPlugin(PaymentPlugin):
    def process(self, amount):
        return f"UPI processed {amount}"

class CardPlugin(PaymentPlugin):
    def process(self, amount):
        return f"Card processed {amount}"

class PaymentGateway:
    def __init__(self):
        self._plugins = {}

    def register_plugin(self, name, plugin):
        self._plugins[name] = plugin

    def pay_with(self, name, amount):
        if name not in self._plugins:
            raise ValueError(f"No plugin registered for '{name}'")
        return self._plugins[name].process(amount)

gateway = PaymentGateway()
gateway.register_plugin("upi", UPIPlugin())
gateway.register_plugin("card", CardPlugin())

print(gateway.pay_with("upi", 4300))
print(gateway.pay_with("card", 4300))

# Explanation: new payment types can be added later by writing a new plugin
# class and registering it, without changing the PaymentGateway class itself.
```

```python
class PositiveAmount:
    """A descriptor that only allows positive numbers to be stored"""
    def __set_name__(self, owner, name):
        self.private_name = "_" + name

    def __get__(self, obj, objtype=None):
        return getattr(obj, self.private_name)

    def __set__(self, obj, value):
        if value <= 0:
            raise ValueError(f"{self.private_name} must be positive, got {value}")
        setattr(obj, self.private_name, value)

class Transfer:
    amount = PositiveAmount()   # descriptor applied to the 'amount' attribute

    def __init__(self, amount):
        self.amount = amount

t = Transfer(4300)
print(f"Transfer amount: {t.amount}")

try:
    t.amount = -50
except ValueError as e:
    print(f"Rejected: {e}")

# Explanation: a descriptor (__get__/__set__) lets us reuse the same
# validation logic across many attributes or classes, not just one property.
```

```python
class Account:
    def __init__(self, account_ref, balance):
        self.account_ref = account_ref
        self.balance = balance
        print(f"Account.__init__ set up {account_ref}")

class AuditedAccount(Account):
    def __init__(self, account_ref, balance, auditor_name):
        super().__init__(account_ref, balance)   # cooperatively calls the parent's init
        self.auditor_name = auditor_name
        print(f"AuditedAccount.__init__ added auditor {auditor_name}")

acc = AuditedAccount("ACC-1099", 22000, "Anjali Das")
print(f"Final object: {acc.account_ref}, balance={acc.balance}, auditor={acc.auditor_name}")

# Explanation: super().__init__(...) lets AuditedAccount reuse Account's setup
# code instead of repeating it, while still adding its own extra attribute.
```

### Q100. Refactor a god object using solid for banking case 3; use sample reference `ACC-1100` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# "Before": one class doing everything (a god object) would mix validation,
# storage, and notifications all together, making it hard to change or test.

# "After": responsibilities are split into small, focused classes (SOLID's
# Single Responsibility Principle)

class TransferValidator:
    def validate(self, amount):
        if amount <= 0:
            raise ValueError("Transfer amount must be positive")

class TransferStore:
    def __init__(self):
        self.saved_transfers = []

    def save(self, transfer_id, amount):
        self.saved_transfers.append((transfer_id, amount))

class TransferNotifier:
    def notify(self, transfer_id):
        print(f"Notification sent for transfer {transfer_id}")

class TransferService:
    """Coordinates the smaller classes instead of doing everything itself"""
    def __init__(self, validator, store, notifier):
        self.validator = validator
        self.store = store
        self.notifier = notifier

    def make_transfer(self, transfer_id, amount):
        self.validator.validate(amount)
        self.store.save(transfer_id, amount)
        self.notifier.notify(transfer_id)

service = TransferService(TransferValidator(), TransferStore(), TransferNotifier())
service.make_transfer("TXN5003", 4300)
print(f"Saved transfers: {service.store.saved_transfers}")

# Explanation: each class now has one job (validate, store, or notify), so
# they can be changed, tested, or replaced independently - this is what
# "refactoring a god object using SOLID" means in practice.
```
