# NumPy Retail Analytics Assignment

## Business Scenario

You are working as a Data Analyst for **SmartMart Retail Stores**. The company operates five stores in different cities and sells six major product categories.

Management wants to analyze sales performance, expenses, profits, targets, discounts, inventory, and store-level performance using NumPy.

You are provided with two main NumPy arrays:

1. `sales_data` — monthly sales revenue generated by each store for different product categories.
2. `cost_data` — total cost incurred by each store for the same product categories.

---

# Part 1: Dataset Creation

````python
import numpy as np

# Store names
stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])

# Product categories
categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])

# Sales revenue in rupees
# Rows represent stores
# Columns represent product categories

sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])

# Cost incurred in rupees
# Rows represent stores
# Columns represent product categories

cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])

print("Stores:")
print(stores)

print("\nProduct Categories:")
print(categories)

print("\nSales Data:")
print(sales_data)

print("\nCost Data:")
print(cost_data)
````

# Part 2: Basic Array Understanding

## Question 1: Array Dimensions

The finance manager wants to confirm how many stores and product categories are present in the data.

Write NumPy code to display:

* Number of dimensions in `sales_data`
* Shape of `sales_data`

* Memory occupied by the array

````python
sales_ndim = sales_data.ndim
sales_shape = sales_data.shape
sales_size = sales_data.size
sales_dtype = sales_data.dtype
sales_nbytes = sales_data.nbytes

print("Number of dimensions:", sales_ndim)
print("Shape:", sales_shape)
print("Total number of values:", sales_size)
print("Data type:", sales_dtype)
print("Memory occupied:", sales_nbytes, "bytes")
````

## Question 3: Sales Data Type Conversion

The reporting application requires sales values in floating-point format.

Convert `sales_data` into:

* `float32`
* `float64`

Compare the memory used by both arrays.

````python
sales_float32 = sales_data.astype(np.float32)
sales_float64 = sales_data.astype(np.float64)

print("float32 array:\n", sales_float32)
print("\nfloat64 array:\n", sales_float64)

print("\nMemory used by int array (original):", sales_data.nbytes, "bytes")
print("Memory used by float32 array:", sales_float32.nbytes, "bytes")
print("Memory used by float64 array:", sales_float64.nbytes, "bytes")
print("float64 uses", sales_float64.nbytes - sales_float32.nbytes, "more bytes than float32")
````

# Business Observations

Based on the complete NumPy analysis above, five key observations emerge:

1. **Mumbai is the standout performer.** It has the highest total sales (₹28,20,000), the highest total profit (₹7,55,000), the best profit margin (26.77%), and is the only store that hit its sales target in all six categories — making it the single store that qualifies for expansion under management's criteria (sales > ₹25,00,000, profit > ₹7,00,000, margin > 25%).

2. **Pune is the clear laggard.** Its total sales (₹19,80,000) and total profit (₹5,05,000) are the lowest of all five stores, and it is the only store that achieved **zero** of its six category targets, despite still posting a positive margin (25.51%) — the issue is under-selling against target, not unprofitability.

3. **Revenue and margin don't move together.** Electronics drives the most revenue by far (₹42,90,000, ~37% of company sales) but has one of the weaker margins (24.24%), while Beauty is the smallest category by revenue (₹10,10,000) yet the most profitable per rupee sold (34.16% margin). SmartMart is volume-heavy in a comparatively low-margin category.

4. **Electronics and Groceries are structural cost problems, not store-specific ones.** These two categories account for 9 of the store-category combinations flagged for cost optimization (cost > ₹2,50,000 and margin < 25%) — nearly every store has both categories flagged, pointing to a company-wide sourcing/logistics issue rather than a single underperforming location.

5. **Store profit margins are tightly clustered (25.5%–26.8%) even though target achievement varies enormously (0 to 6 out of 6 targets met).** This means the target-attainment gap is being driven by category-specific demand or target-setting issues rather than by store-level pricing or cost discipline — every store is running the business efficiently on the sales it does make.

# Management Recommendations

Based on the analysis, three (plus one bonus) actionable recommendations:

1. **Prioritize Mumbai for the next expansion.** It is the only store meeting all three expansion criteria (sales, profit, and margin). Study its category mix and local execution and use it as the template for the next store opening.

2. **Launch a targeted turnaround plan for Pune.** It missed every single category target and trails on both sales and profit. Since its margin is comparable to the other stores, the fix should focus on demand generation (local marketing, footfall, assortment) rather than cost-cutting.

3. **Renegotiate sourcing for Electronics and Groceries company-wide.** Because these two categories are flagged for cost optimization at nearly every store, a centralized vendor/logistics renegotiation is likely to be far more effective than five separate store-level fixes.

4. **Shift promotional spend toward Furniture, Beauty, and Sports.** These categories already clear a 25%+ margin but have below-average sales volume, so incremental marketing rupees spent here should convert to profit more efficiently than in the already-high-volume Electronics category.
