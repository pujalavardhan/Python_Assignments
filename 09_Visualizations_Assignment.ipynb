## 1. Matplotlib

Matplotlib is the most basic plotting library in Python. It lets us make simple charts like line plots and bar charts.

When to use it:
- Simple charts
- When we want full control over the chart
- Basic example below

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

plt.plot(x, y, color='blue', marker='o')
plt.title('Simple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

**Insights:**
- The line goes up as x increases.
- Lowest value is 10 and highest value is 40.
- Overall it shows an increasing trend.

### Customizing the plot

```python
plt.plot(x, y, linestyle='--', color='red', label='Sales')
plt.legend()
plt.show()
```

**Insights:**
- Sales keep increasing from 10 to 40.
- No decrease at any point, so growth is steady.

### Bar chart example

```python
categories = ['Q1', 'Q2', 'Q3', 'Q4']
revenue = [120, 150, 170, 200]

plt.bar(categories, revenue, color='skyblue')
plt.title('Quarterly Revenue')
plt.xlabel('Quarter')
plt.ylabel('Revenue')
plt.show()
```

**Insights:**
- Revenue increases every quarter.
- Q4 has the highest revenue (200).
- Q1 has the lowest revenue (120).

## 2. Seaborn

Seaborn is built on top of Matplotlib. It is used for statistical plots and looks nicer than plain matplotlib with less code.

When to use it:
- For EDA (Exploratory Data Analysis)
- For comparing distributions

```python
import seaborn as sns
import matplotlib.pyplot as plt

flights = sns.load_dataset('flights')

sns.lineplot(data=flights, x='year', y='passengers')
plt.title('Passengers over Time')
plt.show()
```

**Insights:**
- Number of passengers keeps increasing every year.
- It was low in starting years and high in later years.

### Box plot example

```python
tips = sns.load_dataset('tips')

sns.boxplot(data=tips, x='day', y='total_bill')
plt.title('Bill Amount by Day')
plt.show()
```

**Insights:**
- Total bill amount is different for each day.
- Saturday and Sunday have higher bills compared to other days.
- There are some outliers (very high bill values) on some days.

## 3. Plotly

Plotly is used to make interactive charts. We can hover over the chart and zoom in/out.

When to use it:
- When we need interactive charts
- For dashboards

```python
import plotly.express as px

df = px.data.iris()

fig = px.scatter(df, x='sepal_width', y='sepal_length', color='species')
fig.show()
```

**Insights:**
- The scatter plot shows relation between sepal width and sepal length.
- Different species are shown in different colors.
- Setosa species is clearly separate from the other two.

### Line plot example

```python
df2 = px.data.gapminder().query("country == 'India'")

fig = px.line(df2, x='year', y='lifeExp', title='Life Expectancy in India')
fig.show()
```

**Insights:**
- Life expectancy in India keeps increasing over the years.
- It was low around 1952 and much higher by 2007.
- The trend is going up continuously.

## Conclusion

- Matplotlib is good for simple and basic charts.
- Seaborn is good for statistical charts like boxplot with less code.
- Plotly is good for interactive charts where we can hover and zoom.
