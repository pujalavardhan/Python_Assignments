#### 50. Write a Python program that uses multiple built-in functions together to process a list of numbers.

```python
numbers = [12, -5, 8, 0, -3, 15, 7, -9]

print("Original list:", numbers)
print("Length of list:", len(numbers))
print("Sum of list:", sum(numbers))
print("Maximum value:", max(numbers))
print("Minimum value:", min(numbers))
print("Sorted list:", sorted(numbers))
print("Reversed list:", list(reversed(numbers)))
print("Absolute values:", list(map(abs, numbers)))
print("Positive numbers only:", list(filter(lambda x: x > 0, numbers)))
print("Are all values non-zero?", all(numbers))
print("Is any value negative?", any(num < 0 for num in numbers))
```
